### 문제 3-2. 고객 B 한 벡터의 노름과 정규화 복습

#### 문제 설명

교안 **연습문제 4**의 고객 B를 다시 사용합니다. 필수 2의 계산 순서를 익히는 기초 과제입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| B | 6 | 8 |

`x_b=[6,8]`이며 첫 원소는 방문 횟수, 둘째 원소는 장바구니 추가 횟수입니다.

- L1 Norm: 각 원소의 절댓값을 더합니다.
- L2 Norm: 각 원소를 제곱하고, 더한 뒤, 제곱근을 구합니다.
- L2 정규화: 각 원소를 그 벡터의 L2 Norm으로 나눕니다.

#### 요구사항

1. `x_b`를 NumPy 배열로 만들고 출력하세요.
2. `np.linalg.norm(x_b, ord=1)`과 `np.linalg.norm(x_b, ord=2)`로 L1·L2 Norm을 구하여 `b_l1`, `b_l2`에 저장하고 출력하세요.
3. `b_normalized = x_b / b_l2`로 정규화하고 출력하세요.
4. 정규화한 벡터의 L2 Norm을 계산하여 출력하세요. 각 단계의 손계산도 답변란에 적으세요.

#### 작성 도움말

순서는 **원래 벡터 → L1·L2 Norm → L2 Norm으로 나누기 → 새 길이 확인**입니다. 처음부터 공식을 떠올리지 않아도 됩니다. 위 설명과 필수 2-1·2-2를 참고하여 숫자만 대입하세요.

#### 해석 질문

**Q1.** B의 L1 Norm은 얼마인가요? 두 절댓값과 그 합을 차례로 적으세요.

**Q2.** B의 L2 Norm은 얼마인가요? 제곱, 합, 제곱근 순서로 계산하세요.

**Q3.** 정규화할 때 무엇으로 나누나요? 두 원소를 각각 나누어 결과를 적으세요.

**Q4.** 정규화한 벡터의 L2 Norm은 얼마인가요? 길이가 1이라는 것이 두 원소의 합이 1이라는 뜻인가요?

#### 제출 결과

- 원래 벡터와 L1·L2 Norm 출력
- 정규화 벡터와 정규화 후 L2 Norm 출력
- 손계산 과정과 Q1~Q4 답변

````python
# 문제 3-2. 고객 B 한 벡터의 노름과 정규화 복습
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 작성 도움말을 참고하세요.

x_b = np.array([6, 8])

b_l1 = np.linalg.norm(x_b, ord=1)
b_l2 = np.linalg.norm(x_b, ord=2)

print("벡터 b:", x_b)
print("L1 Norm:", b_l1)
print("L2 Norm:", b_l2)

b_normalized = x_b / b_l2
print("B 정규화 값:", b_normalized)

print("B 정규화 후 L2:", np.linalg.norm(b_normalized, ord=2))
````

#### 답변 작성란

**수식 및 계산 과정**

- 작성: 

**Q1. B의 L1 Norm은 얼마인가요? 두 절댓값과 그 합을 차례로 적으세요.**

- 답변: 
    - 첫 원소의 절대값은 |6| = 6, 두 번째 원소의 절대값 |8| = 8
    - 6 + 8 = 14로 계산이 됨

**Q2. B의 L2 Norm은 얼마인가요? 제곱, 합, 제곱근 순서로 계산하세요.**

- 답변: 
    - 첫 원소 6^2 = 36, 두 번째 원소 8^2 = 64 
    - sqrt(36 + 64) = 10

**Q3. 정규화할 때 무엇으로 나누나요? 두 원소를 각각 나누어 결과를 적으세요.**

- 답변: 
    - 정규화할 때 B 벡터의 L2 Norm으로 나눈다.
    - 6 / 10 = 0.6, 8 / 10 = 0.8

**Q4. 정규화한 벡터의 L2 Norm은 얼마인가요? 길이가 1이라는 것이 두 원소의 합이 1이라는 뜻인가요?**

- 답변: 1
    - 아니다. L2 정규화를 해서 길이가 1이 됐다는 것은 벡터의 방향은 유지한 채 길이가 1인 벡터로 바꿨다는 뜻이다. 이는 벡터의 크기 영향을 없애고 뱡향이나 각 성분의 상대적인 비율에 집중하기 위함이다.

|---|---:|---:|---:|
| 인덱스 | 0 | 1 | 2 |
| 점수 | 1 | 2 | 1 |

이 한 상품은 세 종류 중 하나로 분류합니다. 교안 3절의 Argmax를 사용해 가장 높은 확률의 클래스도 확인하세요.

#### 요구사항

#### 작성 도움말

한 벡터용 함수에는 axis=1을 쓰지 않습니다. 가장 큰 점수 2를 빼면 [−1,0,−1]이고, 지수값은 약 [0.3679,1,0.3679]입니다. 이 세 값을 더한 뒤 각각을 같은 합으로 나누세요.

**Q4.** 의류와 가전의 확률이 같은 이유는 무엇인가요?

#### 제출 결과

- 한 벡터용 Softmax 함수, 확률·확률 합
- 예측 인덱스·클래스 이름과 Q1~Q4 답변

````python
# 문제 3-2. 한 상품의 Softmax와 Argmax 복습하기
# 위 데이터와 작성 도움말을 보고 한 단계씩 작성하세요.
# 함수는 문제에 제공된 교안의 코드를 참고하세요.
def softmax(z):
    exp_z = np.exp(z - np.max(z))
    return exp_z / np.sum(exp_z)

z = np.array([1, 2, 1])
classes = np.array(["의류", "식품", "가전"])

p = softmax(z)
print("Softmax 확률:", np.round(p, 3))
print("확률 합:", np.sum(p))

index = np.argmax(p)
print("예측 인덱스:", index)
print("예측 클래스:",classes[index])
````

````python
# 문제 3-1. 새로운 데이터의 우도와 MLE 복습하기
# 1. 문제에 주어진 배열과 값을 작성하세요.
# 2. 요구사항 순서대로 계산하고 출력하세요.
# 3. 아래 답변란에 계산 결과의 뜻을 적으세요.
import numpy as np

y = np.array([1, 1, 0, 1])
candidates = np.array([0.5, 0.75, 0.9])

def likelihood(p, y):
    observed_prob = (p ** y) * ((1 - p) ** (1 - y))
    return np.prod(observed_prob)

values = np.array([likelihood(p, y) for p in candidates])
best_index = np.argmax(values)
best_p = candidates[best_index]

print("구매 수:", np.sum(y))
print("미구매 수:", len(y) - np.sum(y))

print("구매 확률 후보:", candidates)

for p, v in zip(candidates, values):
    print(f"p = {p}: 우도 = {v:.8f}")

print("최대 우도 위치 번호:", best_index)
print("최대 우도 p:", best_p)

p_mle = np.mean(y)
print("MLE:", p_mle)
````

````python
# 문제 3-2. 새로운 고객 데이터의 BCE 복습하기
# 1. 문제에 제시된 배열과 필요한 함수를 작성하세요.
# 2. 요구사항 순서대로 계산하고 출력하세요.
# 3. 아래 답변란에 결과와 그 의미를 적으세요.

import numpy as np

def binary_cross_entropy(y_true, y_prob):
    return -(y_true * np.log(y_prob)
             + (1 - y_true) * np.log(1 - y_prob))

y_true = np.array([1, 0, 1])
p_A = np.array([0.8, 0.3, 0.9])
p_B = np.array([0.6, 0.6, 0.7])

bce_A = binary_cross_entropy(y_true, p_A)
bce_B = binary_cross_entropy(y_true, p_B)

print("모델 A 고객별 BCE:", np.round(bce_A, 3))
print("모델 B 고객별 BCE:", np.round(bce_B, 3))

mean_bce_A = np.mean(bce_A)
mean_bce_B = np.mean(bce_B)

print()
print(f"모델 A 평균 BCE: {mean_bce_A:.3f}")
print(f"모델 B 평균 BCE: {mean_bce_B:.3f}")
````

| x | y | w | b |
|---:|---:|---:|---:|
| 3 | 0 | 0.2 | 0.1 |

p는 여전히 **구매 확률**이고, 미구매 확률은 1−p입니다. 실제 정답이 0이라고 해서 p의 뜻이 미구매 확률로 바뀌지 않습니다.

#### 요구사항

1. 문제 2-1의 제공 함수로 z, p, 1−p, BCE 손실을 계산하세요.
2. 교안의 정리된 식 `dL_dz = p - y`를 사용하세요.
3. `dL_dw = dL_dz * x`, `dL_db = dL_dz * 1`을 계산해 [w,b] 순서로 출력하세요.
4. Q1~Q4에 답하세요. 출력은 소수점 아래 세 자리로 반올림합니다.

#### 작성 도움말

BCE 코드는 `loss = np.logaddexp(0, z) - y * z`입니다. y=0인 경우 손실의 의미는 −ln(1−p)입니다. 이번 과제는 제공된 p−y 공식을 사용하며 BCE와 Sigmoid의 일반 미분식을 다시 유도하거나 별도로 검증할 필요는 없습니다.

````python
# 문제 3-2. 미구매 고객의 기본 Gradient 복습하기
# 1. 제공된 숫자와 코드를 참고해 순전파를 계산하세요.
# 2. 미분식 또는 순서표에 따라 기울기를 계산하세요.
# 3. 결과를 출력하고 아래 질문에 답하세요.

import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

x, y = 3, 0
w, b = 0.2, 0.1

z = w * x + b
p = sigmoid(z)
loss = np.logaddexp(0, z) - y * z

print(f"z: {z:.3f}")
print(f"p: {p:.3f}")
print(f"1 - p: {1-p:.3f}")
print(f"손실: {loss:.3f}")

dL_dz = p - y

print(f"dL/dz: {dL_dz:.3f}")
print(f"p - y: {p - y:.3f}")

dL_dw = dL_dz * x
dL_db = dL_dz * 1

gradient = np.array([dL_dw, dL_db])
print("Gradient:", np.round(gradient, 3))
````

````python
# 문제 3-1. 두 번의 업데이트 복습하기
# 1. 문제의 초기값을 설정하세요.
# 2. 제공 공식에 숫자를 대입하세요.
# 3. 새 값과 결과를 출력하세요.

import numpy as np

x, y = 3, 12
w, b = 2.0, 1.0
lr = 0.02

y_pred = w * x + b
loss = (y - y_pred) ** 2
g_w = 2 * (y_pred - y) * x
g_b = 2 * (y_pred - y)

print("[초기]")
print(f"w: {w:.4f}, b: {b:.4f}")
print(f"예측: {y_pred:.4f}, 손실: {loss:.4f}")
print(f"g_w: {g_w:.4f}, g_b: {g_b:.4f}")

for step in range(1, 3):
    y_pred = w * x + b
    g_w = 2 * (y_pred - y) * x
    g_b = 2 * (y_pred - y)

    w = w - lr * g_w
    b = b - lr * g_b

    y_pred = w * x + b
    loss = (y - y_pred) ** 2

    print(f"\n[{step}회 업데이트 후]")
    print(f"w: {w:.4f}, b: {b:.4f}")
    print(f"예측: {y_pred:.4f}, 손실: {loss:.4f}")
````

- 기울기는 현재 w, b 위치에서의 손실 함수의 순간 변화율이라서, 파라미터가 바뀌면 기울기도 달라진다.
- 첫 업데이트 후 오차가 -5에서 -3으로 줄었으므로 기울기도 [-30, -10]에서 [-18, -6]으로 작아졌다.
- 옛 기울기를 그대로 쓰면 정답에 가까워져도 보폭이 줄지 않아 최솟값을 지나쳐 버릴 수 있다. 따라서 매 업데이트마다 현재 파라미터에서 기울기를 다시 계산해야 한다.
