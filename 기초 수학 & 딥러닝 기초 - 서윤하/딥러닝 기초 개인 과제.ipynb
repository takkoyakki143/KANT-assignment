| 항목 | 세 번째 샘플의 값 |
|---|---:|
| 시멘트량 | 332.5 |
| 물의 양 | 228 |
| 실제 압축강도 | 40.27 |

이어서 이 두 특성을 사용하는 구조 **입력 2 → 은닉 2 → 은닉 2 → 출력 1**을 읽습니다. 각 층은 앞 층의 모든 값과 연결되고 각 계산 뉴런에는 편향 하나가 있습니다.

#### 수식을 말로 읽기

`z = w₁×x₁ + w₂×x₂ + b`에 이번 샘플의 숫자만 바꿔 넣습니다. 가중치와 편향은 동일하게 사용합니다. 파라미터 수는 실습 2-2처럼 연결의 수와 뉴런별 편향의 수를 세어 구합니다.

#### 읽기 도움말 — 두 활동을 구분하세요

먼저 한 뉴런의 계산에 다른 샘플을 넣어 봅니다. 이어서 여러 층을 가진 구조의 가중치·편향 **개수**를 셉니다. 앞에서 구한 z가 뒤에 제시된 전체 신경망의 최종 출력이라는 뜻은 아닙니다. 실습에서 했던 계산 순서와 개수 세기 규칙을 그대로 사용하면 됩니다.

실습 2와 **같은 신경망과 같은 파라미터**를 사용합니다. 이번에는 원본 CSV의 세 번째 샘플인 시멘트량 **332.5**, 물의 양 **228**을 넣습니다. 실제 압축강도는 **40.27**입니다.

| 항목 | 이번 과제의 설정 |
|---|---|
| 입력 순서 | 시멘트량 → 물의 양 |
| W1 | [[0.01, -0.01], [0.00, 0.02]] |
| b1 | [-3.0, 1.0] |
| 은닉 활성화 | ReLU |
| W2 | [[2.0], [1.0]] |
| b2 | [0.5] |
| 출력 활성화 | 따로 적용하지 않음 |

새로운 모델을 설계하거나 성능을 높이는 문제가 아닙니다. 실습 2의 입력만 바꾸고 계산을 다시 따라가세요.

**Q5.** 실제 정답 40.27과 최종 출력은 역할이 어떻게 다른가요? 샘플이 바뀌면서 어떤 값은 바꾸고 어떤 값은 유지했나요?

#### 제출 결과

- 세 번째 입력·정답, Z1_task·H1_task·y_task_hat의 값과 지정된 Shape

- 손계산 및 Q1~Q5 답변


```python
import pandas as pd
import numpy as np

# 실습 준비: CSV 읽기
df = pd.read_csv("concrete_data.csv")
df.columns = df.columns.str.strip()

# 제공 코드: 세 번째 샘플과 실습과 동일한 파라미터입니다.
X_task = df[["cement", "water"]].to_numpy()[2].reshape(1, 2)
y_task_true = df["concrete_compressive_strength"].to_numpy()[2]

print("입력 [시멘트량, 물의 양]:", X_task)
print("정답 [콘크리트 압축 강도]:", y_task_true)

W1 = np.array([[0.01, -0.01],
               [0.00,  0.02]])
b1 = np.array([-3.0, 1.0])
W2 = np.array([[2.0],
               [1.0]])
b2 = np.array([0.5])

# 아래부터 작성하세요.
# 1. Z1_task 계산
Z1_task = X_task @ W1 + b1
print("\nZ1_task:", Z1_task)
print("Z1_task.shape:", Z1_task.shape)

# 2. ReLU로 H1_task 계산
H1_task = np.maximum(0, Z1_task)
print("\nH1_task:", H1_task)
print("H1_task.shape:", H1_task.shape)

# 3. y_task_hat 계산
y_task_hat = H1_task @ W2 + b2
print("\ny_task_hat:", y_task_hat)

# 4. y_task_no_activation 계산
y_task_no_activation = X_task @ W1 + b1
y_task_no_activation = Z1_task @ W2 + b2
print("y_task_no_activation:", y_task_no_activation)

# 5. 값·Shape·np.allclose 비교 결과 출력
print("\n값 비교:")
print("y_task_hat == y_task_no_activation:", np.allclose(y_task_hat, y_task_no_activation))
print("Shape 비교:")
print("y_task_hat.shape:", y_task_hat.shape)
print("y_task_no_activation.shape:", y_task_no_activation.shape)
```

### 문제 3-1. 세 번째 샘플로 한 번의 학습 계산 수행

#### 문제 설명

세 번째 샘플의 시멘트량은 **332.5**, 실제 압축강도는 **40.27**입니다. 계산용 입력은 332.5/100=**3.325**입니다.

이번에도 `w1=0.5, b1=0.1, w2=20, b2=20`에서 출발합니다. **실습에서 수정한 새 파라미터를 이어서 사용하지 않습니다.** 같은 초기 조건에서 다른 샘플에 동일한 방법을 적용하는 복습입니다.

#### 제공된 계산 순서

1. `z1=w1*x+b1`, `h1=ReLU(z1)`, `y_hat=w2*h1+b2`
2. `error=y_hat-y`, `loss=0.5*error**2`
3. `grad_y_hat=error`
4. `grad_w2=grad_y_hat*h1`, `grad_b2=grad_y_hat`
5. `grad_h1=grad_y_hat*w2`, `grad_z1=grad_h1*relu_slope`
6. `grad_w1=grad_z1*x`, `grad_b1=grad_z1`
7. 각 파라미터의 새 값 = 기존 값 - 0.0001 × 해당 기울기
8. 새 파라미터로 순전파와 손실 다시 계산

이번에도 z1이 양수여서 `relu_slope=1.0`을 사용합니다. 모든 미분 식은 실습 1과 같습니다.

#### 읽기 도움말 — 과제의 시작점

입력과 실제 정답은 세 번째 샘플로 바꾸되, 파라미터는 처음 주어진 값으로 다시 시작합니다. 실습에서 얻은 `_new` 값을 가져오지 마세요. 순전파 → 손실 → 역전파 → 한 번 수정 → 다시 계산이라는 순서를 그대로 복습하면 됩니다.

마지막의 반복 곱 계산은 기울기 소실·폭주의 원리 복습입니다. 이 계산의 0.1과 2를 콘크리트 모델의 가중치나 학습률로 넣는 것은 아닙니다.

#### 제출 결과

- 순전파·기울기·새 파라미터·손실 비교 출력
- 별도 기울기 전달 복습 결과
- 손계산 및 Q1~Q5 답변


```python
import pandas as pd
import numpy as np

# 실습 준비: CSV 읽기
df = pd.read_csv("concrete_data.csv")
df.columns = df.columns.str.strip()

# 제공 코드: 원래 초기값으로 다시 시작합니다.
task_x = df["cement"].to_numpy()[2] / 100.0
task_y = df["concrete_compressive_strength"].to_numpy()[2]
task_w1 = 0.5
task_b1 = 0.1
task_w2 = 20.0
task_b2 = 20.0
task_lr = 0.0001

# 아래부터 작성하세요.
# 순전파 → 손실 → 역전파 → 네 파라미터 수정 → 새 손실 확인
# 별도 복습: 1.0 * (0.1 ** 3), 1.0 * (2.0 ** 3)

# 1. 수정 전 순전파
task_z1 = task_w1 * task_x + task_b1
task_h1 = np.maximum(0, task_z1)
task_y_hat = task_w2 * task_h1 + task_b2

# 2. 오차와 손실
task_error = task_y_hat - task_y
task_loss = 0.5 * task_error ** 2

# 3. 역전파 (z1이 양수이므로 relu_slope = 1.0)
task_relu_slope = 1.0
task_grad_y_hat = task_error
task_grad_w2 = task_grad_y_hat * task_h1
task_grad_b2 = task_grad_y_hat
task_grad_h1 = task_grad_y_hat * task_w2
task_grad_z1 = task_grad_h1 * task_relu_slope
task_grad_w1 = task_grad_z1 * task_x
task_grad_b1 = task_grad_z1

# 4. 파라미터 수정 (모두 수정 전 기울기 사용)
task_w1_new = task_w1 - task_lr * task_grad_w1
task_b1_new = task_b1 - task_lr * task_grad_b1
task_w2_new = task_w2 - task_lr * task_grad_w2
task_b2_new = task_b2 - task_lr * task_grad_b2

# 5. 새 파라미터로 순전파와 손실 다시 계산
task_z1_new = task_w1_new * task_x + task_b1_new
task_h1_new = np.maximum(0, task_z1_new)
task_y_hat_new = task_w2_new * task_h1_new + task_b2_new
task_loss_new = 0.5 * (task_y_hat_new - task_y) ** 2

# 출력
print(f"x: {task_x:.4f}, y: {task_y:.4f}")
print()

print("[순전파]")
print(f"z1: {task_z1:.4f}")
print(f"h1: {task_h1:.4f}")
print(f"y_hat: {task_y_hat:.4f}")
print(f"error: {task_error:.4f}")
print(f"loss: {task_loss:.4f}")
print()

print("[역전파]")
print(f"grad_y_hat: {task_grad_y_hat:.4f}")
print(f"grad_h1: {task_grad_h1:.4f}")
print(f"grad_z1: {task_grad_z1:.4f}")
print(f"grad_w1: {task_grad_w1:.4f}")
print(f"grad_b1: {task_grad_b1:.4f}")
print(f"grad_w2: {task_grad_w2:.4f}")
print(f"grad_b2: {task_grad_b2:.4f}")
print()

print("[파라미터 수정 전 → 후]")
print(f"w1: {task_w1:.6f} → {task_w1_new:.6f}")
print(f"b1: {task_b1:.6f} → {task_b1_new:.6f}")
print(f"w2: {task_w2:.6f} → {task_w2_new:.6f}")
print(f"b2: {task_b2:.6f} → {task_b2_new:.6f}")
print()

print("[출력 전 → 후]")
print(f"z1: {task_z1:.4f} → {task_z1_new:.4f}")
print(f"h1: {task_h1:.4f} → {task_h1_new:.4f}")
print(f"y_hat: {task_y_hat:.4f} → {task_y_hat_new:.4f}")
print(f"실제값 y: {task_y:.4f}")
print()

print("[손실 전 → 후]")
print(f"loss: {task_loss:.4f} → {task_loss_new:.4f}")
print("loss_new < loss:", task_loss_new < task_loss)
print()

# 6. 별도 복습: 기울기 전달의 반복 곱
small_3 = 1.0 * (0.1 ** 3)
large_3 = 1.0 * (2.0 ** 3)
print("[기울기 반복 곱 복습]")
print("계수 0.1을 세 번 곱한 결과:", small_3)
print("계수 2를 세 번 곱한 결과:", large_3)
```

lr=0.0001, beta1=0.9, beta2=0.999, eps=1e-8을 사용합니다. 과제는 새 실험이므로 Adam의 m·v는 0, 순번은 t=1로 시작합니다. 실습에서 얻은 theta_adam2·m2·v2를 이어 쓰지 않습니다.

#### 제공된 계산 순서

1. 제공 함수로 초기 출력·손실·기울기를 구합니다.
2. SGD: `theta_sgd = theta0 - lr*g`.
3. Adam: `m = 0.9*m0 + 0.1*g`, `v = 0.999*v0 + 0.001*g**2`.
4. t=1에 맞춰 m과 v를 보정합니다.
5. `theta_adam = theta0 - lr*m_hat/(np.sqrt(v_hat)+eps)`.

#### 읽기 도움말 — 이번 초기화의 이유

Adam 기록을 0으로 만드는 이유는 **독립적인 새 비교 실험**이기 때문입니다. 세 번째 샘플이기 때문만은 아닙니다. 같은 모델의 학습을 계속하면서 다음 샘플이나 묶음으로 넘어가는 경우에는 일반적으로 기록을 이어 사용합니다. 이번 과제는 원래 모델에서 다시 출발하므로 파라미터·기록·순번을 함께 초기 조건으로 맞춥니다.

```python
import pandas as pd
import numpy as np

# 실습 준비 1: 데이터 읽기
df = pd.read_csv("concrete_data.csv")
df.columns = df.columns.str.strip()

# 실습 준비 2: 2장 1강의 순전파·역전파를 묶은 제공 함수
def forward_and_grad(theta, x_value, y_value):
    w1, b1, w2, b2 = theta
    z1 = w1 * x_value + b1
    h1 = np.maximum(0.0, z1)
    prediction = w2 * h1 + b2
    error = prediction - y_value
    loss = 0.5 * error ** 2

    # z1 = 0에서는 코드상 미분값을 0으로 두는 규칙을 사용합니다.
    relu_slope = 1.0 if z1 > 0 else 0.0
    grad_z1 = error * w2 * relu_slope
    gradients = np.array([
        grad_z1 * x_value, 
        grad_z1,
        error * h1,
        error
    ])
    return prediction, loss, gradients

# 과제 제공 코드
task_x = df["cement"].to_numpy()[2] / 100.0
task_y = df["concrete_compressive_strength"].to_numpy()[2]
task_theta0 = np.array([0.5, 0.1, 20.0, 20.0])
task_lr = 0.0001
task_beta1 = 0.9
task_beta2 = 0.999
task_eps = 1e-8
task_t = 1
task_m0 = np.zeros_like(task_theta0)
task_v0 = np.zeros_like(task_theta0)

# 아래부터 작성하세요.
# 초기 기울기를 구한 뒤 두 방식 모두 task_theta0에서 출발합니다.
# SGD 한 번 / Adam 한 번 계산 후 각 출력·손실을 확인하세요.

# 1. 초기 출력·손실·기울기
task_pred0, task_loss0, task_g0 = forward_and_grad(task_theta0, task_x, task_y)

# 2. SGD 한 번 (task_theta0에서 출발)
task_theta_sgd = task_theta0 - task_lr * task_g0
task_pred_sgd, task_loss_sgd, task_g_sgd = forward_and_grad(task_theta_sgd, task_x, task_y)

# 3. Adam 한 번 (task_theta0, m=0, v=0, t=1에서 출발)
task_m1 = task_beta1 * task_m0 + (1 - task_beta1) * task_g0
task_v1 = task_beta2 * task_v0 + (1 - task_beta2) * (task_g0 ** 2)
task_m_hat1 = task_m1 / (1 - task_beta1 ** task_t)
task_v_hat1 = task_v1 / (1 - task_beta2 ** task_t)
task_adam_step = task_lr * task_m_hat1 / (np.sqrt(task_v_hat1) + task_eps)
task_theta_adam = task_theta0 - task_adam_step
task_pred_adam, task_loss_adam, task_g_adam = forward_and_grad(task_theta_adam, task_x, task_y)

# 출력
print("입력 x / 실제 정답 y:", task_x, task_y)
print("파라미터 순서: [w1, b1, w2, b2]")
print()

print("[초기]")
print("초기 파라미터:", task_theta0)
print("초기 출력 / 손실:", round(task_pred0, 8), round(task_loss0, 8))
print("초기 기울기:", np.round(task_g0, 8))
print()

print("[SGD 1회]")
print("뺄 양:", np.round(task_lr * task_g0, 8))
print("SGD 파라미터:", np.round(task_theta_sgd, 8))
print("SGD 출력 / 손실:", round(task_pred_sgd, 8), round(task_loss_sgd, 8))
print()

print("[Adam 1회]")
print("m1:", np.round(task_m1, 8))
print("v1:", np.round(task_v1, 8))
print("m_hat1:", np.round(task_m_hat1, 8))
print("v_hat1:", np.round(task_v_hat1, 8))
print("뺄 양:", np.round(task_adam_step, 10))
print("Adam 파라미터:", np.round(task_theta_adam, 10))
print("Adam 출력 / 손실:", round(task_pred_adam, 8), round(task_loss_adam, 8))
print()

print("[수정 전후 손실 비교]")
print("초기 손실:", round(task_loss0, 8))
print("SGD 1회 후 손실:", round(task_loss_sgd, 8), "/ 감소:", task_loss_sgd < task_loss0)
print("Adam 1회 후 손실:", round(task_loss_adam, 8), "/ 감소:", task_loss_adam < task_loss0)
```

합성곱은 필터 길이 3, stride 1, padding 없음, bias 0입니다. 풀링은 크기 2, stride 2, padding 없음입니다.

```python
import pandas as pd
import numpy as np

# 실습 준비 1: 데이터 읽기
df = pd.read_csv("jena_climate_2009_2016.csv", nrows=16)
times = pd.to_datetime(df["Date Time"], format="%d.%m.%Y %H:%M:%S")
x = df["T (degC)"].iloc[:8].to_numpy(dtype=float)
x_home = df["T (degC)"].iloc[8:16].to_numpy(dtype=float)

# 실습 준비 2: 제공 함수
def conv1d_valid(values, kernel):
    results = []
    for i in range(len(values) - len(kernel) + 1):
        window = values[i:i + len(kernel)]
        value = np.sum(window * kernel)
        results.append(value)
    return np.array(results)


def pool_pairs(values):
    maximums = []
    averages = []
    for i in range(0, len(values), 2):
        window = values[i:i + 2]
        maximums.append(np.max(window))
        averages.append(np.mean(window))
    return np.array(maximums), np.array(averages)

print("합성곱 함수와 풀링 함수 준비 완료")

# 과제: 실습에서 사용한 코드의 입력을 x_home으로 바꿔 적용하세요.
# 실습과 같은 계산을 x_home에 적용
home_filter = np.array([-1, 0, 1])
home_feature = conv1d_valid(x_home, home_filter)
home_relu = np.maximum(0.0, home_feature)
home_max, home_avg_unused = pool_pairs(home_relu)

# 네 단계의 값과 길이 출력
print("입력:", x_home, "길이:", len(x_home))
print("합성곱:", home_feature.round(4), "길이:", len(home_feature))
print("ReLU:", home_relu.round(4), "길이:", len(home_relu))
print("Max Pooling:", home_max.round(4), "길이:", len(home_max))

# 첫 합성곱 값 손계산 확인
print(f"\n첫 합성곱 = (-1)*{x_home[0]} + 0*{x_home[1]} + 1*{x_home[2]} = {home_feature[0]:.4f}")

# 세 풀링 묶음 확인
for i in range(0, len(home_relu), 2):
    pair = home_relu[i:i + 2]
    print(f"묶음 {i//2 + 1}: {pair.round(4)} → max={np.max(pair):.4f}")
```

```python
import pandas as pd
import numpy as np

# 실습 준비 1: 데이터 읽기 (기온을 10으로 나눈 값 사용)
df = pd.read_csv("jena_climate_2009_2016.csv", nrows=12)
times = pd.to_datetime(df["Date Time"], format="%d.%m.%Y %H:%M:%S")
x = df["T (degC)"].iloc[:6].to_numpy(dtype=float) / 10.0
x_home = df["T (degC)"].iloc[6:12].to_numpy(dtype=float) / 10.0

# 실습 준비 2: RNN 제공 함수
def run_rnn(values, reset_each=False):
    h = 0.0
    rows = []
    for t, value in enumerate(values, start=1):
        h_prev = 0.0 if reset_each else h
        z = 0.5 * value + 0.5 * h_prev
        h = np.tanh(z)
        rows.append({"t": t, "x": value, "h_prev": h_prev, "z": z, "h": h})
    return pd.DataFrame(rows)

# 실습 준비 3: LSTM 제공 함수
def sigmoid(value):
    return 1.0 / (1.0 + np.exp(-value))


def run_lstm(values):
    h = 0.0
    c = 0.0
    rows = []
    for t, value in enumerate(values, start=1):
        h_prev = h
        c_prev = c
        s = 0.5 * value + 0.2 * h_prev
        f = sigmoid(s + 1.0)
        i = sigmoid(s)
        g = np.tanh(s)
        o = sigmoid(s)
        keep = f * c_prev
        add = i * g
        c = keep + add
        h = o * np.tanh(c)
        rows.append({"t": t, "x": value, "h_prev": h_prev,
                     "f": f, "i": i, "g": g, "o": o,
                     "c_prev": c_prev, "keep": keep, "add": add,
                     "c": c, "h": h})
    return pd.DataFrame(rows)

# 과제: 동일한 제공 함수에 과제 입력을 넣으세요.
home_rnn = run_rnn(x_home)       # 매 시점 초기화 옵션 사용 안 함
home_lstm = run_lstm(x_home)     # 새 구간이므로 h, c 모두 0에서 시작

print("[과제 RNN]")
print(home_rnn[["t", "x", "h"]].round(6).to_string(index=False))
print("\n[과제 LSTM]")
print(home_lstm[["t", "x", "c", "h"]].round(6).to_string(index=False))

# 첫 시점 대입 확인
rnn_h1 = np.tanh(0.5 * x_home[0])
r1 = home_lstm.iloc[0]
lstm_c1 = r1["i"] * r1["g"]      # c0 = 0이므로 c1 = i1*g1
print(f"\n첫 RNN h = tanh(0.5*{x_home[0]:.3f}) = {rnn_h1:.6f}")
print(f"첫 LSTM c = i1*g1 = {r1['i']:.6f}*{r1['g']:.6f} = {lstm_c1:.6f}")

# 마지막 상태
print("\n마지막 RNN h:", round(home_rnn["h"].iloc[-1], 6))
print("마지막 LSTM c:", round(home_lstm["c"].iloc[-1], 6))
print("마지막 LSTM h:", round(home_lstm["h"].iloc[-1], 6))
```

단계별 계산·크기, 행별 가중치 합, 첫 행 대입 계산, Q1~Q4 답변. 모델 학습이나 심화 비교는 요구하지 않습니다.

```python
import pandas as pd
import numpy as np

# 실습 준비 1: 데이터와 가중치
df = pd.read_csv("jena_climate_2009_2016.csv", nrows=6)
times = pd.to_datetime(df["Date Time"], format="%d.%m.%Y %H:%M:%S")
raw = df[["T (degC)", "rh (%)"]].to_numpy(dtype=float)
X_all = raw / np.array([10.0, 100.0])
X = X_all[:3]
X_home = X_all[3:6]

WQ = np.array([[1.0, 0.0], [0.0, 1.0]])
WK = np.array([[1.0, 0.0], [0.0, 0.5]])
WV = np.array([[1.0], [0.0]])

# 실습 준비 2: 행별 Softmax 제공 함수
def softmax_rows(scores):
    shifted = scores - np.max(scores, axis=1, keepdims=True)
    exp_scores = np.exp(shifted)
    return exp_scores / np.sum(exp_scores, axis=1, keepdims=True)

print("행별 Softmax 함수 준비 완료")

# 과제: X_home에 같은 가중치 행렬을 적용하세요.
# X_home에 같은 가중치 행렬 적용
Q_home = X_home @ WQ
K_home = X_home @ WK
V_home = X_home @ WV

# 내적, 크기 조정, Softmax, 가중합
d_home = K_home.shape[1]
scores_home = Q_home @ K_home.T
scaled_home = scores_home / np.sqrt(d_home)
weights_home = softmax_rows(scaled_home)
output_home = weights_home @ V_home

# 각 값과 크기 출력
print("Q_home:\n", np.round(Q_home, 6), "\nshape:", Q_home.shape)
print("\nK_home:\n", np.round(K_home, 6), "\nshape:", K_home.shape)
print("\nV_home:\n", np.round(V_home, 6), "\nshape:", V_home.shape)
print("\nweights_home:\n", np.round(weights_home, 6), "\nshape:", weights_home.shape)
print("\nweights_home 행별 합:", np.round(weights_home.sum(axis=1), 6))
print("\noutput_home:\n", np.round(output_home, 6), "\nshape:", output_home.shape)

# 첫 Query의 첫 내적 점수 대입 확인
s11 = Q_home[0, 0] * K_home[0, 0] + Q_home[0, 1] * K_home[0, 1]
print(f"\n첫 내적 = {Q_home[0,0]:.3f}*{K_home[0,0]:.3f} + {Q_home[0,1]:.3f}*{K_home[0,1]:.4f} = {s11:.6f}")

# 첫 가중합 출력 대입 확인
terms = " + ".join(f"{weights_home[0,j]:.6f}*{V_home[j,0]:.3f}" for j in range(3))
print(f"첫 출력 = {terms} = {output_home[0,0]:.6f}")

# 첫 Query에서 가장 큰 비중의 관측 시각
best = np.argmax(weights_home[0])
print("\n첫 Query 가중치:", np.round(weights_home[0], 6))
print("가장 큰 비중:", round(weights_home[0, best], 6), "→ 관측 시각:", times.iloc[3 + best])
```

## 4장 2강 Transformer 아키텍쳐 과제

#### 문제 설명

준비 코드의 X_home은 00:40·00:50·01:00의 기온·상대습도 입력입니다. 실습과 같은 위치 표현 함수와 두 헤드 가중치로 계산하세요.

**이 과제는 별도의 입력 구간을 처리하는 독립 실습**입니다. 따라서 위치 번호는 다시 0·1·2로 시작합니다. 앞 구조 문제의 미래 예측을 실행하는 것이 아니며, 이 과제에서는 세 관측값을 이미 받은 입력으로 취급합니다.

#### 요구사항

1. `PE_home=position_encoding(len(X_home))`을 구하고 `X_home_pos=X_home+PE_home`을 계산하세요.
2. 같은 제공 함수와 가중치로 `A1_home,H1_home`, `A2_home,H2_home`을 구하세요.

5. 첫 행의 위치 덧셈과 WO 출력 변환을 손계산하세요.

#### 작성 도움말

실습 코드의 X_pos를 X_home_pos로 바꾸어 적용하면 됩니다. WQ·WK·WV·WO는 바꾸지 않습니다. `np.concatenate([H1_home,H2_home],axis=1)`을 사용하세요.

#### 해석 질문

**Q1.** PE_home이 실습 PE와 같은 이유는 무엇인가요? 첫 X_home_pos 행은 무엇인가요?

**Q2.** 두 헤드 출력·연결 결과·최종 출력의 크기는 각각 무엇인가요? 각 Attention 가중치의 행 합은 얼마인가요?

**Q3.** 첫 위치의 두 헤드 출력과 M_home 첫 행은 무엇인가요? WO를 이용한 대입 과정을 쓰세요.

**Q4.** 실습과 출력 숫자가 달라진 이유는 무엇인가요? M_home을 미래 기온 예측값이라고 부를 수 있나요?

#### 제출 결과

단계별 배열·크기, 가중치 행 합, 첫 행 계산, Q1~Q4 답변. 추가 학습이나 심화 구현은 요구하지 않습니다.

```python
import pandas as pd
import numpy as np

# 실습 준비 1: 데이터
df = pd.read_csv("jena_climate_2009_2016.csv", nrows=6)
times = pd.to_datetime(df["Date Time"], format="%d.%m.%Y %H:%M:%S")
raw = df[["T (degC)", "rh (%)"]].to_numpy(dtype=float)
X_all = raw / np.array([10.0, 100.0])
X = X_all[:3]
X_home = X_all[3:6]

# 실습 준비 2: 위치 정보와 Attention 제공 함수
def position_encoding(length):
    positions = np.arange(length, dtype=float)
    return np.column_stack([np.sin(positions), np.cos(positions)])


def softmax_rows(scores):
    shifted = scores - np.max(scores, axis=1, keepdims=True)
    exp_scores = np.exp(shifted)
    return exp_scores / np.sum(exp_scores, axis=1, keepdims=True)


def attention_head(values, wq, wk, wv):
    q = values @ wq
    k = values @ wk
    v = values @ wv
    d_k = k.shape[1]
    scores = q @ k.T / np.sqrt(d_k)
    weights = softmax_rows(scores)
    head_output = weights @ v
    return weights, head_output

print("위치 정보와 Attention 함수 준비 완료")

# 실습 준비 3: 헤드별 가중치와 출력 가중치
WQ1 = np.array([[1.0], [0.0]])
WK1 = np.array([[1.0], [0.0]])
WV1 = np.array([[1.0], [0.0]])
WQ2 = np.array([[0.0], [1.0]])
WK2 = np.array([[0.0], [1.0]])
WV2 = np.array([[0.0], [1.0]])
WO = np.array([[1.0, 0.5], [0.5, 1.0]])

print("각 헤드 가중치 크기:", WQ1.shape, WQ2.shape)
print("출력 가중치 WO:", WO)

# 새 구간의 위치 정보를 구하고 더하기 (위치는 다시 0, 1, 2)
PE_home = position_encoding(len(X_home))
X_home_pos = X_home + PE_home

# 같은 가중치로 두 헤드 계산
A1_home, H1_home = attention_head(X_home_pos, WQ1, WK1, WV1)
A2_home, H2_home = attention_head(X_home_pos, WQ2, WK2, WV2)

# 연결 후 WO 곱하기
C_home = np.concatenate([H1_home, H2_home], axis=1)
M_home = C_home @ WO

# 단계별 값과 크기, 각 헤드의 행 합 출력
print("X_home_pos:\n", np.round(X_home_pos, 6), "\nshape:", X_home_pos.shape)
print("\nA1_home 행별 합:", np.round(A1_home.sum(axis=1), 6))
print("A2_home 행별 합:", np.round(A2_home.sum(axis=1), 6))
print("\nH1_home:\n", np.round(H1_home, 6), "\nshape:", H1_home.shape)
print("\nH2_home:\n", np.round(H2_home, 6), "\nshape:", H2_home.shape)
print("\nC_home:\n", np.round(C_home, 6), "\nshape:", C_home.shape)
print("\nM_home:\n", np.round(M_home, 6), "\nshape:", M_home.shape)

# 첫 행 위치 덧셈 손계산 확인
print(f"\n첫 행: {np.round(X_home[0], 6)} + {np.round(PE_home[0], 6)} = {np.round(X_home_pos[0], 6)}")

# 첫 행 WO 출력 변환 손계산 확인
h1, h2 = H1_home[0, 0], H2_home[0, 0]
print(f"M_home[0,0] = {h1:.6f}*1 + {h2:.6f}*0.5 = {h1 + 0.5*h2:.6f}")
print(f"M_home[0,1] = {h1:.6f}*0.5 + {h2:.6f}*1 = {0.5*h1 + h2:.6f}")
```

#### 제출 결과

두 설정 비교 표·복원 검증 손실·네 기법 역할 정리, Q1~Q3 답변.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 실습 준비 1: 데이터 분리와 표준화
data_path = Path("concrete_data.csv")
if not data_path.exists():
    data_path = Path("concrete_data(1).csv")
original = pd.read_csv(data_path)
original.columns = original.columns.str.strip()
df = original.drop_duplicates().reset_index(drop=True)
target = "concrete_compressive_strength"
features = [name for name in df.columns if name != target]
indices = np.random.default_rng(42).permutation(len(df))
train_df = df.iloc[indices[:80]].copy()
valid_df = df.iloc[indices[80:280]].copy()

train_raw = train_df[features].to_numpy(dtype=float)
valid_raw = valid_df[features].to_numpy(dtype=float)
x_mean = train_raw.mean(axis=0)
x_std = train_raw.std(axis=0)
X_train = (train_raw - x_mean) / x_std
X_valid = (valid_raw - x_mean) / x_std
y_raw = train_df[[target]].to_numpy(dtype=float)
y_valid_raw = valid_df[[target]].to_numpy(dtype=float)
y_mean = y_raw.mean()
y_std = y_raw.std()
y_train = (y_raw - y_mean) / y_std
y_valid = (y_valid_raw - y_mean) / y_std

print("원본 / 중복 제거 후:", original.shape, df.shape)
print("학습 / 검증 / 미사용 행:", len(train_df), len(valid_df), len(df)-280)

# 실습 준비 2: 학습·예측 제공 함수
def predict(values, params):
    w1, b1, w2, b2 = params
    return np.tanh(values @ w1 + b1) @ w2 + b2


def train_demo():
    rng = np.random.default_rng(7)
    w1 = rng.normal(0.0, 0.3, size=(8, 32))
    b1 = np.zeros((1, 32))
    w2 = rng.normal(0.0, 0.1, size=(32, 1))
    b2 = np.zeros((1, 1))
    records, checkpoints = [], []
    for epoch in range(1, 1201):
        hidden = np.tanh(X_train @ w1 + b1)
        prediction = hidden @ w2 + b2
        grad_output = 2.0 * (prediction - y_train) / len(X_train)
        grad_w2 = hidden.T @ grad_output
        grad_b2 = grad_output.sum(axis=0, keepdims=True)
        grad_hidden = (grad_output @ w2.T) * (1.0 - hidden**2)
        grad_w1 = X_train.T @ grad_hidden
        grad_b1 = grad_hidden.sum(axis=0, keepdims=True)
        w1 -= 0.03 * grad_w1
        b1 -= 0.03 * grad_b1
        w2 -= 0.03 * grad_w2
        b2 -= 0.03 * grad_b2
        params = (w1, b1, w2, b2)
        train_loss = np.mean((predict(X_train, params) - y_train)**2)
        valid_loss = np.mean((predict(X_valid, params) - y_valid)**2)
        records.append({"epoch": epoch, "train_loss": train_loss, "valid_loss": valid_loss})
        checkpoints.append(tuple(value.copy() for value in params))
    return pd.DataFrame(records), checkpoints

# 실습 준비 3: Early Stopping 판단 제공 함수
def replay_early_stopping(history, checkpoints, patience):
    best_loss = float("inf")
    best_epoch = None
    best_params = None
    wait = 0
    trace = []
    stopped = False
    for row in history.itertuples(index=False):
        improved = row.valid_loss < best_loss
        if improved:
            best_loss = row.valid_loss
            best_epoch = int(row.epoch)
            best_params = tuple(p.copy() for p in checkpoints[best_epoch - 1])
            wait = 0
        else:
            wait += 1
        trace.append({"epoch": int(row.epoch), "valid_loss": row.valid_loss,
                      "improved": improved, "wait": wait, "best_epoch": best_epoch})
        if wait >= patience:
            stopped = True
            break
    result = {"stopped": stopped, "last_checked_epoch": trace[-1]["epoch"],
              "best_epoch": best_epoch, "best_valid_loss": best_loss}
    return result, pd.DataFrame(trace), best_params

# 과제에 필요한 실습 결과 (실습 1-1, 2-2)
history, checkpoints = train_demo()
result20, trace20, best_params = replay_early_stopping(history, checkpoints, patience=20)
print("patience=20 중단 / 최적 epoch:", result20["last_checked_epoch"], result20["best_epoch"])

# 과제
result10, trace10, params10 = replay_early_stopping(history, checkpoints, patience=10)
loss10 = np.mean((predict(X_valid, params10) - y_valid) ** 2)

# patience=20의 복원 검증 손실 (실습 2-2에서 계산하던 값)
restored_loss = np.mean((predict(X_valid, best_params) - y_valid) ** 2)

print(f"patience=10 복원 검증 손실: {loss10:.6f}")

# patience 20 vs 10 비교 표
compare = pd.DataFrame([
    {"patience": 20, "중단 여부": result20["stopped"],
     "마지막 확인 epoch": result20["last_checked_epoch"],
     "최적 epoch": result20["best_epoch"],
     "최저 검증 손실": result20["best_valid_loss"],
     "복원 검증 손실": restored_loss},
    {"patience": 10, "중단 여부": result10["stopped"],
     "마지막 확인 epoch": result10["last_checked_epoch"],
     "최적 epoch": result10["best_epoch"],
     "최저 검증 손실": result10["best_valid_loss"],
     "복원 검증 손실": loss10},
])
print(compare.round(6).to_string(index=False))

print("\n[patience=10 판단 기록 마지막 12행]")
print(trace10.tail(12).round(6).to_string(index=False))
```

**Q3. 네 기법의 역할과 평가 때의 동작을 각각 정리하세요. 이들을 모두 켜면 반드시 가장 좋은 모델이 되나요?**

- 내 답변: 
    - Dropout: 학습 중 일부 뉴런의 활성값을 무작위로 0으로 만들어 특정 뉴런에 대한 과도한 의존을 줄인다. 평가할 때는 Dropout을 적용하지 않는다.
    - Weight Decay: 가중치 업데이트에 축소 항을 추가하여 가중치가 지나치게 커지는 것을 억제한다. 평가할 때는 학습된 가중치를 그대로 사용한다.
    - Batch Normalization: 학습 중 배치의 평균과 분산을 이용해 활성값을 정규화하고, gamma와 beta로 조정한다. 평가할 때는 학습 과정에서 저장한 이동평균과 이동 분산을 사용한다.
    - Early Stopping: 검증 손실이 일정 횟수 동안 개선되지 않으면 학습을 중단하고 최적 시점의 가중치를 복원한다. 평가할 때는 복원된 가중치를 사용한다.
    - 네 기법을 모두 적용한다고 해서 반드시 가장 좋은 모델이 되는 것은 아니다. 각 기법의 효과는 데이터와 모델, 하이퍼파라미터에 따라 달라지므로 실제 검증 성능을 비교해야 한다.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 공통 준비 1: 데이터 분리와 표준화
data_path = Path("concrete_data.csv")
if not data_path.exists():
    data_path = Path("concrete_data(1).csv")
original = pd.read_csv(data_path)
original.columns = original.columns.str.strip()
df = original.drop_duplicates().reset_index(drop=True)
target = "concrete_compressive_strength"
features = [name for name in df.columns if name != target]
indices = np.random.default_rng(42).permutation(len(df))
train_df = df.iloc[indices[:80]].copy()
valid_df = df.iloc[indices[80:280]].copy()

train_raw = train_df[features].to_numpy(dtype=float)
valid_raw = valid_df[features].to_numpy(dtype=float)
x_mean = train_raw.mean(axis=0)
x_std = train_raw.std(axis=0)
X_train = (train_raw - x_mean) / x_std
X_valid = (valid_raw - x_mean) / x_std
y_raw = train_df[[target]].to_numpy(dtype=float)
y_valid_raw = valid_df[[target]].to_numpy(dtype=float)
y_mean = y_raw.mean()
y_std = y_raw.std()
y_train = (y_raw - y_mean) / y_std
y_valid = (y_valid_raw - y_mean) / y_std

print("원본 / 중복 제거 후:", original.shape, df.shape)
print("학습 / 검증 / 미사용 행:", len(train_df), len(valid_df), len(df)-280)

# 공통 준비 2: 학습·예측 제공 함수
def predict(values, params):
    w1, b1, w2, b2 = params
    return np.tanh(values @ w1 + b1) @ w2 + b2


def train_demo(lr=0.03, epochs=1200):
    rng = np.random.default_rng(7)
    w1 = rng.normal(0.0, 0.3, size=(8, 32))
    b1 = np.zeros((1, 32))
    w2 = rng.normal(0.0, 0.1, size=(32, 1))
    b2 = np.zeros((1, 1))
    records = []
    for epoch in range(1, epochs + 1):
        hidden = np.tanh(X_train @ w1 + b1)
        prediction = hidden @ w2 + b2
        grad_output = 2.0 * (prediction - y_train) / len(X_train)
        grad_w2 = hidden.T @ grad_output
        grad_b2 = grad_output.sum(axis=0, keepdims=True)
        grad_hidden = (grad_output @ w2.T) * (1.0 - hidden**2)
        grad_w1 = X_train.T @ grad_hidden
        grad_b1 = grad_hidden.sum(axis=0, keepdims=True)
        w1 -= lr * grad_w1
        b1 -= lr * grad_b1
        w2 -= lr * grad_w2
        b2 -= lr * grad_b2
        params = (w1, b1, w2, b2)
        train_loss = np.mean((predict(X_train, params) - y_train)**2)
        valid_loss = np.mean((predict(X_valid, params) - y_valid)**2)
        records.append({"epoch": epoch, "train_loss": train_loss,
                        "valid_loss": valid_loss})
        if not np.isfinite([train_loss, valid_loss]).all():
            print("유한하지 않은 손실이 발생해 기록을 중단했습니다:", epoch)
            break
    return pd.DataFrame(records)

# 과제에 필요한 실습 결과 (실습 1-2)
history = train_demo(lr=0.03, epochs=1200)

# 과제

candidates = history[history["epoch"].isin([100, 584, 1200])].copy()
candidates["gap"] = candidates["valid_loss"] - candidates["train_loss"]

train_choice = int(candidates.loc[candidates["train_loss"].idxmin(), "epoch"])
valid_choice = int(candidates.loc[candidates["valid_loss"].idxmin(), "epoch"])
gap_choice = int(candidates.loc[candidates["gap"].idxmin(), "epoch"])

print(candidates.round(6).to_string(index=False))
print(f"\n학습 손실 최소 후보: epoch {train_choice}")
print(f"검증 손실 최소 후보: epoch {valid_choice}")
print(f"gap 최소 후보: epoch {gap_choice}")
```

- 내 답변:

고정 여부 확인 표, 두 조건의 손실 비교 표, Q1~Q3 답변.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 준비 1: 파일 연결 → 기간 분리 → Source 기준 전처리
def find_file(names):
    for name in names:
        if Path(name).exists():
            return Path(name)
    raise FileNotFoundError("노트북과 같은 폴더에 파일을 넣어 주세요: " + ", ".join(names))


power = pd.read_csv(find_file(["전기사용량(2).csv", "전기사용량.csv"]), encoding="utf-8-sig")
building = pd.read_csv(find_file(["building_info(2).csv", "building_info.csv"]), encoding="utf-8-sig")
power.columns = power.columns.str.strip()
building.columns = building.columns.str.strip()
assert not power.duplicated(["건물번호", "일시"]).any()
assert building["건물번호"].is_unique
data = power.merge(building[["건물번호", "건물유형", "연면적(m2)", "냉방면적(m2)"]],
                   on="건물번호", how="left", validate="many_to_one", indicator=True)
assert data["_merge"].eq("both").all()
data["time"] = pd.to_datetime(data["일시"], format="%Y%m%d %H")
data["hour"] = data["time"].dt.hour
data["weekday"] = data["time"].dt.dayofweek
data = data.sort_values(["건물번호", "time"]).reset_index(drop=True)

source = data[(data["건물번호"].isin([1,2,3,4])) &
              (data["time"] >= "2022-06-01") & (data["time"] < "2022-07-01")].copy()
target_train = data[(data["건물번호"] == 5) &
                    (data["time"] >= "2022-07-01") & (data["time"] < "2022-07-08")].copy()
target_valid = data[(data["건물번호"] == 5) &
                    (data["time"] >= "2022-07-08") & (data["time"] < "2022-07-15")].copy()
features = ["기온(C)", "습도(%)", "hour", "weekday", "연면적(m2)", "냉방면적(m2)"]
target = "전력소비량(kWh)"

fill_values = source[features].median()
source_values = source[features].fillna(fill_values).to_numpy(dtype=float)
x_mean = source_values.mean(axis=0)
x_std = source_values.std(axis=0)
x_std = np.where(x_std == 0, 1.0, x_std)
y_mean = float(source[target].mean())
y_std = float(source[target].std(ddof=0))


def transform(frame):
    values = frame[features].fillna(fill_values).to_numpy(dtype=float)
    X = (values - x_mean) / x_std
    y = (frame[[target]].to_numpy(dtype=float) - y_mean) / y_std
    return X, y


X_source, y_source = transform(source)
X_train, y_train = transform(target_train)
X_valid, y_valid = transform(target_valid)
assert np.isfinite(X_source).all() and np.isfinite(X_train).all() and np.isfinite(X_valid).all()
print("사전학습 / 대상 학습 / 대상 검증:", len(source), len(target_train), len(target_valid))

# 준비 2: 모델·학습 제공 함수
def copy_model(params):
    return {name: value.copy() for name, value in params.items()}


def init_model(seed=7):
    rng = np.random.default_rng(seed)
    return {"W1": rng.normal(0.0, 0.25, (6, 8)), "b1": np.zeros((1, 8)),
            "W2": rng.normal(0.0, 0.1, (8, 1)), "b2": np.zeros((1, 1))}


def hidden_features(X, params):
    return np.tanh(X @ params["W1"] + params["b1"])


def predict_scaled(X, params):
    return hidden_features(X, params) @ params["W2"] + params["b2"]


def train_model(start, X, y, lr, epochs, train_backbone):
    params = copy_model(start)
    records = []
    for epoch in range(1, epochs + 1):
        H = hidden_features(X, params)
        prediction = H @ params["W2"] + params["b2"]
        grad_output = 2.0 * (prediction - y) / len(X)
        grad_W2 = H.T @ grad_output
        grad_b2 = grad_output.sum(axis=0, keepdims=True)
        if train_backbone:
            grad_H = (grad_output @ params["W2"].T) * (1.0 - H**2)
            grad_W1 = X.T @ grad_H
            grad_b1 = grad_H.sum(axis=0, keepdims=True)
            params["W1"] -= lr * grad_W1
            params["b1"] -= lr * grad_b1
        params["W2"] -= lr * grad_W2
        params["b2"] -= lr * grad_b2
        loss = float(np.mean((predict_scaled(X, params) - y)**2))
        records.append({"epoch": epoch, "train_loss": loss})
    return params, pd.DataFrame(records)


def parameter_changes(before, after):
    return pd.DataFrame([
        {"parameter": name, "count": before[name].size,
         "max_abs_change": float(np.max(np.abs(after[name] - before[name])))}
        for name in ["W1", "b1", "W2", "b2"]
    ])


def target_scores(models):
    rows = []
    for name, params in models.items():
        rows.append({"model": name,
                     "target_train_mse": float(np.mean((predict_scaled(X_train, params) - y_train)**2)),
                     "target_valid_mse": float(np.mean((predict_scaled(X_valid, params) - y_valid)**2))})
    return pd.DataFrame(rows)

# 과제에 필요한 실습 결과 (실습 1-1, 1-2, 2-1)
initial = init_model()
pretrained, source_history = train_model(
    initial, X_source, y_source, lr=0.03, epochs=600, train_backbone=True)

feature_start = copy_model(pretrained)
feature_start["W2"] = np.random.default_rng(21).normal(0.0, 0.1, (8, 1))
feature_start["b2"] = np.zeros((1, 1))
feature_model, feature_history = train_model(
    feature_start, X_train, y_train, lr=0.03, epochs=300, train_backbone=False)

fine_model, fine_history = train_model(
    feature_model, X_train, y_train, lr=0.003, epochs=100, train_backbone=True)
print("실습 모델 준비 완료: pretrained, feature_model, fine_model")

# 과제
# 1. 같은 출발점에서 은닉층 고정한 채 추가 학습
frozen_more, frozen_history = train_model(
    feature_model, X_train, y_train,
    lr=0.003, epochs=100, train_backbone=False
)

# 2. 은닉층 고정 여부 확인
frozen_changes = parameter_changes(feature_model, frozen_more)
print("[feature_model → frozen_more 파라미터 변화]")
print(frozen_changes.to_string(index=False))
print("W1 동일:", np.array_equal(feature_model["W1"], frozen_more["W1"]))
print("b1 동일:", np.array_equal(feature_model["b1"], frozen_more["b1"]))

# 3. 두 조건의 대상 손실 비교
homework_scores = target_scores({"Frozen continuation": frozen_more,
                                 "Fine-tuning": fine_model})
print("\n[두 조건의 대상 손실 비교]")
print(homework_scores.to_string(index=False))

# 4. 업데이트 대상 개수
frozen_updated = frozen_more["W2"].size + frozen_more["b2"].size
fine_updated = sum(value.size for value in fine_model.values())
print("\nFrozen continuation 업데이트 수:", frozen_updated)
print("Fine-tuning 업데이트 수:", fine_updated)
```

**Q2.** 상황 B의 선택과 이유는 무엇인가요? 가장 빠른 전략이라고 해서 성능 기준을 무시해도 되나요?

**Q3.** 상황 C에서는 무엇을 선택하나요? B와 선택이 달라질 수 있는 이유와 이번 비교의 한계를 설명하세요.

#### 제출 결과

통합 비교 표, 조건을 통과한 전략, A·B·C의 선택과 Q1~Q3 답변.


```python
from pathlib import Path
from time import perf_counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 공통 준비 1: 데이터 연결·분리·고정 단위 변환
def find_file(names):
    for name in names:
        if Path(name).exists():
            return Path(name)
    raise FileNotFoundError("노트북과 같은 폴더에 파일을 넣어 주세요: " + ", ".join(names))


power = pd.read_csv(find_file(["전기사용량(2).csv", "전기사용량.csv"]), encoding="utf-8-sig")
building = pd.read_csv(find_file(["building_info(2).csv", "building_info.csv"]), encoding="utf-8-sig")
power.columns = power.columns.str.strip()
building.columns = building.columns.str.strip()
assert not power.duplicated(["건물번호", "일시"]).any()
assert building["건물번호"].is_unique
data = power.merge(building[["건물번호", "건물유형", "연면적(m2)", "냉방면적(m2)"]],
                   on="건물번호", how="left", validate="many_to_one", indicator=True)
assert data["_merge"].eq("both").all()
data["time"] = pd.to_datetime(data["일시"], format="%Y%m%d %H")
data["hour"] = data["time"].dt.hour
data["weekday"] = data["time"].dt.dayofweek
data = data.sort_values(["건물번호", "time"]).reset_index(drop=True)

source = data[(data["건물번호"].isin([1,2,3,4])) &
              (data["time"] >= "2022-06-01") & (data["time"] < "2022-07-01")].copy()
target_train = data[(data["건물번호"] == 5) &
                    (data["time"] >= "2022-07-01") & (data["time"] < "2022-07-08")].copy()
target_valid = data[(data["건물번호"] == 5) &
                    (data["time"] >= "2022-07-08") & (data["time"] < "2022-07-15")].copy()
features = ["기온(C)", "습도(%)", "hour", "weekday", "연면적(m2)", "냉방면적(m2)"]
target = "전력소비량(kWh)"

x_center = np.array([20.0, 50.0, 11.5, 3.0, 0.0, 0.0])
x_scale = np.array([10.0, 20.0, 7.0, 2.0, 100000.0, 100000.0])
y_scale = 1000.0


def transform(frame):
    X = (frame[features].to_numpy(dtype=float) - x_center) / x_scale
    y = frame[[target]].to_numpy(dtype=float) / y_scale
    return X, y


X_source, y_source = transform(source)
X_train, y_train = transform(target_train)
X_valid, y_valid = transform(target_valid)
for array in [X_source, y_source, X_train, y_train, X_valid, y_valid]:
    assert np.isfinite(array).all(), "선택 구간의 결측값과 숫자 형식을 확인하세요."
print("Source / Target Train / Target Validation:", len(source), len(target_train), len(target_valid))

# 공통 준비 2: 모델·학습·전략 비교·시간 측정 제공 함수
def copy_model(params):
    return {name: value.copy() for name, value in params.items()}


def init_model(seed=7):
    rng = np.random.default_rng(seed)
    return {"W1": rng.normal(0.0, 0.25, (6,8)), "b1": np.zeros((1,8)),
            "W2": rng.normal(0.0, 0.1, (8,1)), "b2": np.zeros((1,1))}


def hidden_features(X, params):
    return np.tanh(X @ params["W1"] + params["b1"])


def predict_scaled(X, params):
    return hidden_features(X, params) @ params["W2"] + params["b2"]


def fit_segment(start, X, y, lr, epochs, freeze, X_check=None, y_check=None):
    params = copy_model(start)
    records = []
    fixed_H = hidden_features(X, params) if freeze else None
    fixed_check = hidden_features(X_check, params) if freeze and X_check is not None else None
    for epoch in range(1, epochs+1):
        H = fixed_H if freeze else hidden_features(X, params)
        pred = H @ params["W2"] + params["b2"]
        grad_output = 2.0 * (pred - y) / len(X)
        grad_W2 = H.T @ grad_output
        grad_b2 = grad_output.sum(axis=0, keepdims=True)
        if not freeze:
            grad_hidden = (grad_output @ params["W2"].T) * (1.0 - H**2)
            grad_W1 = X.T @ grad_hidden
            grad_b1 = grad_hidden.sum(axis=0, keepdims=True)
            params["W1"] -= lr * grad_W1
            params["b1"] -= lr * grad_b1
        params["W2"] -= lr * grad_W2
        params["b2"] -= lr * grad_b2
        current_H = fixed_H if freeze else hidden_features(X, params)
        train_pred = current_H @ params["W2"] + params["b2"]
        row = {"epoch": epoch, "train_mse": float(np.mean((train_pred-y)**2))}
        if X_check is not None:
            H_check = fixed_check if freeze else hidden_features(X_check, params)
            valid_pred = H_check @ params["W2"] + params["b2"]
            row["valid_mse"] = float(np.mean((valid_pred-y_check)**2))
        records.append(row)
    return params, pd.DataFrame(records)


def pretrain_source():
    started = perf_counter()
    params, history = fit_segment(init_model(), X_source, y_source,
        lr=0.03, epochs=600, freeze=False)
    return params, history, perf_counter()-started


def make_start(strategy, pretrained):
    params = init_model() if strategy == "Scratch" else copy_model(pretrained)
    params["W2"] = np.random.default_rng(21).normal(0.0, 0.1, (8,1))
    params["b2"] = np.zeros((1,1))
    return params


strategies = ["Scratch", "Feature extraction", "Fine-tuning"]


def run_strategy(strategy, pretrained):
    if strategy not in strategies:
        raise ValueError("알 수 없는 전략: " + strategy)
    started = perf_counter()
    start = make_start(strategy, pretrained)
    first, h1 = fit_segment(start, X_train, y_train, lr=0.03, epochs=300,
        freeze=(strategy != "Scratch"), X_check=X_valid, y_check=y_valid)
    final, h2 = fit_segment(first, X_train, y_train, lr=0.003, epochs=100,
        freeze=(strategy == "Feature extraction"), X_check=X_valid, y_check=y_valid)
    h2["epoch"] += 300
    history = pd.concat([h1,h2], ignore_index=True)
    return final, history, perf_counter()-started, first


def benchmark(pretrained, repeats=3):
    measured = []
    for repeat in range(repeats):
        _, _, source_seconds = pretrain_source()
        measured.append({"repeat": repeat+1, "stage": "Source pretraining", "seconds": source_seconds})
        order = strategies[repeat % 3:] + strategies[:repeat % 3]
        for name in order:
            _, _, seconds, _ = run_strategy(name, pretrained)
            measured.append({"repeat": repeat+1, "stage": name, "seconds": seconds})
    return pd.DataFrame(measured)

# 과제에 필요한 실습 결과 (실습 1-1, 1-2, 2-1, 2-2)
pretrained, source_history, source_seconds = pretrain_source()

models, histories, mid_models = {}, {}, {}
for name in strategies:
    model, history, elapsed, middle = run_strategy(name, pretrained)
    models[name] = model
    histories[name] = history
    mid_models[name] = middle

rows = []
for name, model in models.items():
    train_mse = float(np.mean((predict_scaled(X_train, model) - y_train) ** 2))
    valid_mse = float(np.mean((predict_scaled(X_valid, model) - y_valid) ** 2))
    rows.append({"strategy": name, "train_mse": train_mse, "valid_mse": valid_mse})
performance = pd.DataFrame(rows)

timing_runs = benchmark(pretrained, repeats=3)
median_times = timing_runs.groupby("stage")["seconds"].median()
counts = {"Scratch": (65, 65), "Feature extraction": (9, 9), "Fine-tuning": (9, 65)}
rows = []
for name in strategies:
    target_sec = float(median_times[name])
    source_sec = 0.0 if name == "Scratch" else float(median_times["Source pretraining"])
    rows.append({"strategy": name,
                 "updated_first_300": counts[name][0],
                 "updated_last_100": counts[name][1],
                 "target_seconds": target_sec,
                 "source_seconds": source_sec,
                 "estimated_total_seconds": target_sec + source_sec})
efficiency = pd.DataFrame(rows)
print("실습 결과 준비 완료: performance, efficiency")

# 과제
# 1. 성능·효율 표 연결
comparison = performance.merge(efficiency, on="strategy", validate="one_to_one")
print(comparison.to_string(index=False))

# 2. 성능 조건 통과 후보
eligible = comparison[comparison["valid_mse"] <= 1.60]
print("\n[valid_mse <= 1.60 통과 전략]")
print(eligible[["strategy", "valid_mse", "target_seconds", "estimated_total_seconds"]].to_string(index=False))

# 3. 상황별 선택
choice_a = comparison.loc[comparison["valid_mse"].idxmin()]
print(f"\nA (검증 MSE 최소): {choice_a['strategy']} | valid_mse={choice_a['valid_mse']:.4f}")

if eligible.empty:
    print("B, C: 1.60 조건을 통과한 전략이 없습니다.")
else:
    choice_b = eligible.loc[eligible["target_seconds"].idxmin()]
    choice_c = eligible.loc[eligible["estimated_total_seconds"].idxmin()]
    print(f"B (조건 통과 + 대상 학습 시간 최소): {choice_b['strategy']} | "
          f"valid_mse={choice_b['valid_mse']:.4f}, target_seconds={choice_b['target_seconds']:.4f}")
    print(f"C (조건 통과 + Source 포함 추정 시간 최소): {choice_c['strategy']} | "
          f"valid_mse={choice_c['valid_mse']:.4f}, estimated_total_seconds={choice_c['estimated_total_seconds']:.4f}")
```
