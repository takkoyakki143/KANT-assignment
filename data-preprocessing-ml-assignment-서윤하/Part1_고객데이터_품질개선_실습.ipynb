## 문제 2. 분석 가능한 고객 데이터로 정제

### 업무 상황

품질 진단 결과를 바탕으로 고객 데이터를 분석 가능한 상태로 정리해야 합니다.  
처리 과정에서 고객의 실제 구매 행동 정보가 불필요하게 손실되지 않도록 데이터 특성을 고려하세요.

### 요구사항

1. 원본 데이터를 보존한 상태에서 정제용 데이터를 생성하세요.
2. 분석에 맞지 않는 수치형·날짜형 자료형과 범주 표기를 정리하세요.
3. 결측치와 전체 행 기준 중복 데이터를 처리하세요.
4. 주요 수치형 컬럼의 이상치를 탐지하고 적절한 방법으로 처리하세요.
5. 처리 전후의 결측치, 중복값과 이상치 상태를 확인하세요.

### 힌트

- 변환할 수 없는 문자열은 결측치로 바꾼 뒤 일관되게 처리할 수 있습니다.
- 이상치는 무조건 삭제하기보다 실제 우수 고객의 행동일 가능성도 고려하세요.
- IQR은 이상치 후보를 확인하는 대표적인 방법입니다.

```python
# TODO: 원본을 보존하고 정제용 데이터 df_clean을 생성하세요.
df_clean = df.copy()

# TODO: 수치형·날짜형 자료형과 범주 표기를 정리하세요.
df_clean['Age'].unique()
df_clean['Age'] = df_clean['Age'].replace('thirty', '30')
df_clean['Age'] = df_clean['Age'].str.replace('세', '', regex=False)
df_clean['Age'] = pd.to_numeric(df_clean['Age'], errors = 'coerce')
df_clean['Age'] = df_clean['Age'].astype('Int64')

df_clean['AveragePurchaseAmount'].unique()
df_clean['AveragePurchaseAmount'] = (df_clean['AveragePurchaseAmount'].str.replace(',', '', regex=False))
df_clean['AveragePurchaseAmount'] = pd.to_numeric(df_clean['AveragePurchaseAmount'], errors = 'coerce')
df_clean['AveragePurchaseAmount'] = df_clean['AveragePurchaseAmount'].astype('float')

df_clean['LastPurchaseDate'] = pd.to_datetime(df_clean['LastPurchaseDate'], format = 'mixed', errors = 'coerce')
df_clean['SignupDate'] = pd.to_datetime(df_clean['SignupDate'], format = 'mixed', errors = 'coerce')

df_clean['Gender'].unique()
df_clean['Gender'] = df_clean['Gender'].str.strip().str.lower()
df_clean['Gender'] = df_clean['Gender'].replace({
    'm':'Male',
    'male':'Male',
    'female':'Female',
    'f':'Female',
    'other' : 'Other',
    'nan' : np.nan
})

df_clean['Region'].unique()
df_clean['Region'] = df_clean['Region'].str.strip().str.lower()
df_clean['Region'] = df_clean['Region'].replace({
    'gyeonggi' : 'Gyeonggi',
    'gwangju': 'Gwangju',
    'daegu' : 'Daegu',
    'daejeon' : 'Daejeon',
    'busan' : 'Busan',
    'seoul' : 'Seoul',
    'incheon' : 'Incheon',
    'nan' : np.nan
})

df_clean['MembershipLevel'].unique()
df_clean['MembershipLevel'] = df_clean['MembershipLevel'].str.strip().str.lower()
df_clean['MembershipLevel'] = df_clean['MembershipLevel'].replace({
    'silver' : 'Silver',
    'basic' : 'Basic',
    'vip' : 'VIP',
    'gold' : 'Gold',
    'nan' : np.nan
})

df_clean['CouponUsed'].unique()
df_clean['CouponUsed'] = df_clean['CouponUsed'].str.strip().str.lower()
df_clean['CouponUsed'] = df_clean['CouponUsed'].replace({
    'n' : 'N',
    'no' : 'N',
    'y' : 'Y',
    'yes' : 'Y'
})

df_clean['PreferredCategory'].unique()
df_clean['PreferredCategory'] = df_clean['PreferredCategory'].str.strip().str.lower()
df_clean['PreferredCategory'] = df_clean['PreferredCategory'].replace({
    'beauty' : 'Beauty',
    'home' : 'Home',
    'fashion' : 'Fashion',
    'food' : 'Food',
    'electronics' : 'Electronics',
    'sports' : 'Sports',
    'nan' : np.nan
})
```

```python
# TODO: 결측치와 전체 행 기준 중복 데이터를 처리하세요.
# Gender, Region, MembershipLevel, LastPurchaseDate, PreferredCategory는 결측치를 임의로 채우거나 전체 행을 삭제하는 것보다는 유지 

# Age 결측치 -> 극단적인 이상치가 존재하기 때문에 중앙값(median)으로 대체
age_median = df_clean['Age'].median()
df_clean['Age'] = df_clean['Age'].fillna(age_median)

# Average Purchase Amount 결측치 -> 극단적인 이상치는 발견되지 않았기 때문에 평균(mean)으로 대체
avg_mean = df_clean['AveragePurchaseAmount'].mean()
df_clean['AveragePurchaseAmount'] = df_clean['AveragePurchaseAmount'].fillna(avg_mean)

# Satisfaction Socre 결측치 -> 극단적인 이상치는 발견되지 않았기 때문에 평균(mean)으로 대체
score_mean = df_clean['SatisfactionScore'].mean()
df_clean['SatisfactionScore'] = df_clean['SatisfactionScore'].fillna(score_mean)
df_cleaned = df_clean.drop_duplicates()

# TODO: 처리 전후 결과를 확인하세요.
print('처리 전 데이터 크기:', df_clean.shape)
print('처리 후 데이터 크기:', df_cleaned.shape)
```

```python
def get_iqr_bounds(series):
    # TODO: Q1, Q3, IQR, 하한과 상한을 반환하세요.
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    return lower_bound, upper_bound

# TODO: 주요 수치형 컬럼의 이상치를 탐지하고 처리하세요.
display(df_cleaned.describe())
print()

print(get_iqr_bounds(df_cleaned['Age']))
outliers_age = df_cleaned[(df_cleaned['Age'] < 9.5) | (df_cleaned['Age'] > 69.5)]
print(outliers_age['Age'])
df_cleaned_iqr = df_cleaned[df_cleaned['Age'] <= 105] # 이상치 제거
print()

print(get_iqr_bounds(df_cleaned['AveragePurchaseAmount']))
outliers_avg = df_cleaned[(df_cleaned['AveragePurchaseAmount'] < -16425) | (df_cleaned['AveragePurchaseAmount'] > 250575)]
print(outliers_avg['AveragePurchaseAmount'])
print()

print(get_iqr_bounds(df_cleaned['SatisfactionScore']))
outliers_score = df_cleaned[(df_cleaned['SatisfactionScore'] < 1.7) | (df_cleaned['SatisfactionScore'] > 5.7)]
print(outliers_score['SatisfactionScore'])
print()

print(get_iqr_bounds(df_cleaned['TotalPurchaseAmount']))
outliers_total = df_cleaned[(df_cleaned['TotalPurchaseAmount'] < -338050) | (df_cleaned['TotalPurchaseAmount'] > 1059950)]
print(outliers_total['TotalPurchaseAmount'])
print()
# 극단적으로 높은 총 구매금액들이 몇 건 있었지만, 해당 변수는 누적 금액이기 때문에 회원 가입일과 최근 구매일을 함께 확인하였다.
# 일부 고액 구매 고객은 장기간 이용 이력이 확인되었으며, 3천만원의 구매 기록을 가진 고객은 약 1년 간 쓴 금액이지만 현실적으로 발생 불가능한
# 금액이라고 보기 어려워 이상치로 판단하지 않았다.

print(get_iqr_bounds(df_cleaned['VisitCount']))
outliers_visit = df_cleaned[(df_cleaned['VisitCount'] < 1) | (df_cleaned['VisitCount'] > 17)]
print(outliers_visit['VisitCount'])
print()
# 총 구매금액과 마찬가지로 극단적으로 높은 값들이 몇 건 있었으나 회원 가입일과 최근 구매일을 비교하였을 때 일부 높은 방문 횟수의 고객은 장기 이용 고객임을 알아냈다.
# 또한 약 6개월의 기간동안 170회 방문한 고객도 단순히 서비스를 굉장히 자주 이용하는 고객일 수 있기 때문에 이상치로 판단하지 않았다.

```

```python
# TODO: 처리 전후 이상치 상태를 확인하세요.
def count_outliers(df, column):
    lower, upper = get_iqr_bounds(df[column])

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    return len(outliers)

numeric_cols = [
    'Age',
    'AveragePurchaseAmount',
    'SatisfactionScore',
    'TotalPurchaseAmount',
    'VisitCount'
]

print('처리 전 이상치 수')
for col in numeric_cols:
    print(col, ':', count_outliers(df_clean, col))

print()

print('처리 후 이상치 수')
for col in numeric_cols:
    print(col, ':', count_outliers(df_cleaned_iqr, col))

print('Age의 비현실적인 고연령 값을 제거하는 과정에서 해당 행 전체가 제외되었기 때문에 다른 수치형 변수의 분포도 일부 변화하였다. ' \
'이로 인해 AveragePurchaseAmount의 IQR 기준이 재계산되면서 이상치 개수가 소폭 증가했으나, 새로운 이상치가 생성된 것은 아니며 분포 변화에 따른 기준선 변화로 판단하였다.')
```
